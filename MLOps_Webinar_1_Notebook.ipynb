# Why We Use WSL 2 for MLOps

**WSL 2 (Windows Subsystem for Linux)** provides a Linux environment on Windows, making it easier to use the tools and workflows commonly used in production MLOps. It also provides better alignment with Linux-based cloud environments and simplifies containerization and scripting.

| **Feature** | **Native Windows** | **WSL 2** |
|---|---|---|
| **MLOps Tool Support** | Limited; may require workarounds | **Broad Linux-native compatibility** |
| **Docker Integration** | Additional configuration may be required | **Efficient WSL 2 integration** |
| **Environment Parity** | Lower | **Closer to typical cloud environments** |
| **Paths & Scripting** | `\`, PowerShell | **`/`, Bash** |



> **Webinar Note:** Our hands-on MLOps webinar uses **WSL 2** as the standard environment so that all participants follow a consistent, Linux-aligned setup. **You may ignore the WSL setup instructions if you already have WSL configured, are running a Linux operating system, or are using macOS.**

````

Download:

```bash
curl -L -o diabetes.csv \
https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv
```

Verify:

```bash
ls -lh diabetes.csv
```

The dataset should be available at:

```text
data/raw/diabetes.csv
```

> 💡 We will explore the dataset and understand its features, target, structure, and data quality during **Phase 1 — Understand the Data**.



## Phase 1 — Data Loading & Validation

In this phase, We will:

1. Identify the dataset and its structure
2. Understand the features and target
3. Inspect the data types and values
4. Look for missing or unusual values
5. Understand the distribution of the target variable

### Question to keep in mind

> **Is our data ready to be given to a machine-learning algorithm?**


##  Phase 2 — Data Preparation

Our data is understood and cleaned. Now we need to prepare it for machine learning.

In this phase, we will:

1. Separate **features (`X`)** from the **target (`y`)**
2. Split the data into **training and testing sets**
3. Prepare the numerical features for modelling
4. Build a reusable preprocessing pipeline

>  **Goal:** Transform our cleaned dataset into data that can be safely given to ML models.

## Phase 3 — Train & Compare Models

Our data is prepared and our preprocessing component is ready.

Now we will train four different machine-learning models:

1. **Logistic Regression**
2. **Random Forest**
3. **XGBoost**
4. **MLP Neural Network**

For each model, we will follow the same workflow:

```text
Training Data
     ↓
Preprocessing
     ↓
Model
     ↓
Training
     ↓
Evaluation
```
We are **not studying the algorithms here**.

> Our objective is to run comparable experiments and identify a candidate champion that we can take into the MLOps lifecycle.

## Phase 4 — Experiment Tracking with MLflow

We have completed our initial model experimentation and selected a candidate champion.

However, the results currently exist only in our notebook.

In a real ML workflow, we need to track:

- Which model was trained
- Which parameters were used
- What metrics were achieved
- Which model artifact was produced

**MLflow** provides the experiment-tracking infrastructure needed to manage this information.

Start the MLflow UI in a terminal:

```bash
mlflow ui --backend-store-uri sqlite:///mlflow.db --port 5000
```

Use the UI to compare the candidate runs and identify the champion.


## 5. Model Selection, Registration & Serving

Select the best model and register it as `DiabetesPredictionChampion`.
The champion model is now registered in the MLflow Model Registry.

In this phase, we will serve the registered model so that it can receive
prediction requests independently of the training workflow.

The model will be loaded directly from the MLflow Model Registry.


Serve the registered model:

```bash
mlflow models serve \
  -m "models:/DiabetesPredictionChampion/3" \
  -p 5001 \
  --host 127.0.0.1 \
  --env-manager local
```


## 6. Model Packaging

Export the registered model for independent use.

## Model Packaging Flow

```
MLflow Model Registry
        ↓
DiabetesPredictionChampion
        ↓
Version 3
        ↓
Export Model Artifact
        ↓
deployment/model_export/
        ↓
Run Inference
```


## Run Prediction Using the Packaged Model

The packaged champion model has been loaded successfully from:

`deployment/model_export`

We will now use this independently loaded model to generate predictions.

This is different from Phase 5, where predictions were generated using the model served through the MLflow Model Server.

Here, the prediction happens directly through the packaged model object.

The notebook will call a reusable prediction function from:

`deployment/predict_packaged_model.py`

The function will:

1. receive input data
2. generate predictions using the packaged model
3. return the predictions

### Key Question

> Can the exported model actually perform inference independently of the MLflow Model Registry and Model Server?

## 7. Docker Containerization

Docker helps package the application together with its runtime environment.

## Our Goal

In this phase, we will:

1. Define the runtime environment required by our model.
2. Create a Docker image for model inference.
3. Build the Docker image.
4. Verify that the image was created successfully.
5. Run the model inside a Docker container.
6. Send an inference request to the container.
7. Verify that the containerized model produces predictions.

## Why Docker?

A Docker container gives us an isolated and reproducible environment in which our model can run.

Instead of saying:

> "Install Python, MLflow, scikit-learn and all the required dependencies, then run the model."

we can package the environment into a Docker image.

The deployment flow becomes:

```text
Docker Image
     ↓
Docker Container
     ↓
Model Runtime
     ↓
Prediction
```
Create a new file under the project directory, name it Dockerfile and insert the following content:

**Dockerfile**

```dockerfile
FROM python:3.12-slim
WORKDIR /app
COPY deployment/model_export /app/model_export
RUN pip install --no-cache-dir -r /app/model_export/requirements.txt
EXPOSE 5001
CMD ["mlflow", "models", "serve", "-m", "/app/model_export", "--host", "0.0.0.0", "--port", "5001", "--env-manager", "local"]
```

Build:

```bash
docker build -t diabetes-model:v1 .
```

Verify:

```bash
docker images diabetes-model
```

Run:

```bash
docker run --rm -p 5001:5001 diabetes-model:v1
```

Check the model server is running by opening a new terminal and running the following command: 

```bash
curl http://127.0.0.1:5001/ping
```

Send a prediction request: 

```bash
curl -X POST http://127.0.0.1:5001/invocations \
  -H "Content-Type: application/json" \
  -d '{
    "dataframe_split": {
      "columns": [
        "Pregnancies",
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI",
        "DiabetesPedigreeFunction",
        "Age"
      ],
      "data": [
        [6, 148, 72, 35, 0, 33.6, 0.627, 50]
      ]
    }
  }'
```


## 8. Batch Inference

### What is Batch Inference?

Batch inference means taking a collection of input records and generating predictions for all of them together.

Instead of:

```text
One record
    ↓
Model
    ↓
One prediction

we process:

Many records
    ↓
Model
    ↓
Many predictions

We can also generate predictions for the complete input file for batch inference.

## Implement Batch Inference

We will now run the packaged champion model against a batch of records.

For this webinar, the original dataset is used to simulate an incoming batch of data.

In a production system, this input would typically come from a newly generated file, database table, data lake, or another data source.

The batch inference implementation is kept in:

deployment/batch_inference.py

Output: `data/processed/batch_predictions.csv`


# Phase 9 — Airflow Orchestration

## Understand Airflow Orchestration

What is Airflow?

Apache Airflow is a workflow orchestration platform.

It allows us to define:

What tasks need to run
The order in which they should run
Dependencies between tasks
When workflows should run
What should happen when a task succeeds or fails

Our Airflow DAG

We will represent our MLOps workflow as a DAG:

```
prepare_data
      ↓
train_models
      ↓
evaluate_models
      ↓
select_best_model
      ↓
package_model
      ↓
run_inference
```

DAG stands for:

Directed Acyclic Graph

In simple terms:

A DAG is a collection of tasks with defined dependencies and a defined execution direction.

## Map MLOps Components to Airflow Tasks

Each Airflow task will call an existing Python implementation.

| Airflow Task | Existing Implementation | Responsibility |
|---|---|---|
| `prepare_data` | `src/features/prepare_data.py` | Prepare training data |
| `train_models` | `src/models/train_models.py` | Train candidate models |
| `evaluate_models` | `src/models/evaluate_models.py` | Evaluate trained models |
| `select_best_model` | `src/tracking/select_champion.py` | Select champion model |
| `package_model` | `deployment/export_champion.py` | Export champion model |
| `run_inference` | `deployment/batch_inference.py` | Generate batch predictions |

Airflow will therefore coordinate our existing components rather than duplicate their implementation.

### Final automated workflow

```text
prepare_data
      ↓
train_models
      ↓
evaluate_models
      ↓
select_best_model
      ↓
package_model
      ↓
run_inference
```

Start Airflow:

```bash
airflow standalone
```

Trigger the DAG:

```bash
airflow dags trigger diabetes_mlops_pipeline
```


## Takeaway

**Experiment → Track → Select → Register → Package → Infer → Orchestrate**
